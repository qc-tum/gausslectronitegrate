# Example of evaluating nuclear overlap integrals of Gausslet orbitals

```python
# see the README for build and install instructions of this package
import gausslectronitegrate as gli
```

```python
# Gausslet G10 coefficients
import sys
sys.path.append("../gausslets/")
from gausslets import get_gausslet_g10_coeffs
b_list, _ = get_gausslet_g10_coeffs()
```

```python
# define a Cartesian grid by tuples of the form (istart, num points) in each coordinate direction
grid = [
    (-1, 3),  # x
    ( 0, 2),  # y
    (-2, 4),  # z
]
```

```python
# overall number of grid points
gli.grid_num_points(grid)
```

```python
# nuclear configuration
nuclear_positions = [(-0.2, 1.3, 0.5), (0.7, 0, -2.1)]
nuclear_charges = [2, 1]
```

```python
# truncation tolerance
# (values close to zero imply higher accuracy but longer computation time)
tol = 1e-16
```

```python
# evaluate nuclear overlap integrals for Gausslet orbitals centered at the grid points
ngi = gli.compute_nuclear_gausslet_integrals(b_list, grid, nuclear_positions, nuclear_charges, tol)
```

```python
# function returns a NumPy matrix of size "number of grid points"
print(type(ngi))
print(ngi.shape)
```

```python
# check symmetry
import numpy as np
print(np.linalg.norm(ngi.T - ngi))
```

```python
# linear index of origin
iorigin = gli.grid_point_to_linear_index(grid, [0, 0, 0])
iorigin
```

```python
# show the overlap integral value for both Gausslets centered at the origin
ngi[iorigin, iorigin]
```

```python
# compute the individual overlap integral for both Gausslets centered at the origin
gli.compute_nuclear_gausslet_integral(b_list, ([0, 0, 0], [0, 0, 0]), nuclear_positions, nuclear_charges, tol)
```

```python
# evaluate the overlap integral for two other points, as an example
point_a = [-1,  0, -2]
point_b = [ 0,  1,  1]
gli.compute_nuclear_gausslet_integral(b_list, (point_a, point_b), nuclear_positions, nuclear_charges, tol)
```

```python
# compare with value in 'ngi' array
ia = gli.grid_point_to_linear_index(grid, point_a)
ib = gli.grid_point_to_linear_index(grid, point_b)
ngi[ia, ib]
```
