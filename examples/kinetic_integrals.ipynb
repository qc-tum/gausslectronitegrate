# Example of evaluating kinetic overlap integrals of Gausslet orbitals

```python
# see the README for build and install instructions of this package
import gausslectronitegrate as gli
```

```python
# Gausslet G10 coefficients
from gausslet_coefficients import get_gausslet_g10_coeffs
b_list, _ = get_gausslet_g10_coeffs()
```

```python
# define a Cartesian grid by tuples of the form (istart, num points) in each coordinate direction
grid = [
    (-1, 3),  # x
    (-2, 5),  # y
    (-2, 4),  # z
]
```

```python
# overall number of grid points
gli.grid_num_points(grid)
```

```python
# evaluate kinetic overlap integrals for Gausslet orbitals centered at the grid points
kgi = gli.compute_kinetic_gausslet_integrals(b_list, grid)
```

```python
# function returns a NumPy matrix of size "number of grid points"
print(type(kgi))
print(kgi.shape)
```

```python
# show the overlap integral value for both Gausslets centered at the same (arbitrary) grid point
kgi[7, 7]
```
